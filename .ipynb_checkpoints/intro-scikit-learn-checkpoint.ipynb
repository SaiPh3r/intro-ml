<h1 style="color: #8000ff; font-size: 3rem; text-align: center; font-family: 'Segoe UI', sans-serif; margin-bottom: 20px;">

````python
# fill the make column
cars_missing_data["Make"].fillna("missing" , inplace = True)
# fill the colour column
cars_missing_data["Colour"].fillna("missing" , inplace = True)
#fill the odometor column
cars_missing_data["Odometer (KM)"].fillna(cars_missing_data["Odometer (KM)"].mean() , inplace = True)
#fill door column
cars_missing_data["Doors"].fillna(4 , inplace = True)
````

````python
cars_missing_data.isna().sum()
````

````python
# lets remove rows which price is na
cars_missing_data.dropna(inplace=True)
````

````python
cars_missing_data.isna().sum()
````

````python
X = cars_missing_data.drop("Price" , axis=1)
Y = cars_missing_data["Price"]
````

````python
# turn the categories in number
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
categorical_featues = ["Make" , "Colour" , "Doors"]
one_hot = OneHotEncoder()
transformer = ColumnTransformer([("one_hot" , one_hot , categorical_featues )] , remainder="passthrough")

transformed_X = transformer.fit_transform(cars_missing_data)
transformed_X
````

<h1 style="color: #8000ff; font-size: 3rem; text-align: center; font-family: 'Segoe UI', sans-serif; margin-bottom: 20px;">
   Option2 : fill missing data with Scikit-Learn
</h1>

````python
car_missing_sk = pd.read_csv("scikit-learn-data/car-sales-extended-missing-data.csv")
car_missing_sk.head()
car_missing_sk.isna().sum()
````

````python
#drop the rows with no labels
car_missing_sk.dropna(subset=["Price"] , inplace= True)
car_missing_sk.isna().sum()
````

````python
x = car_missing_sk.drop("Price",axis = 1)
y = car_missing_sk["Price"]
````

````python
# fill missing values using scikit learn 
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

# fill categorical values with missing and numerical values with mean
category_imputer = SimpleImputer(strategy="constant" , fill_value="missing")
door_imputer = SimpleImputer(strategy="constant" , fill_value = 4)
num_imputer = SimpleImputer(strategy="mean")

#define columns
category_features = ["Make","Colour"]
door_features = ["Doors"]
num_features = ["Odometer (KM)"]

#create an imputer (something that fills value)
imputer = ColumnTransformer([("category_imputer" , category_imputer  , category_features) , ("door_imputer" , door_imputer , door_features),
                            ("num_imputer" , num_imputer , num_features)])
#transform the data
filled_x = imputer.fit_transform(x)
filled_x
````

````python
cars_updated_data = pd.DataFrame(filled_x , columns=["Make" , "colour", "Doors", "Odometer (km) "])
````

````python
cars_updated_data.isna().sum()
````

````python
# lets convert the data to numerical data
# turn the categories in number
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
categorical_features = ["Make" , "colour", "Doors"]
one_hot = OneHotEncoder()
transformer = ColumnTransformer([("one_hot", one_hot, categorical_features)], remainder="passthrough")


transformed_x = transformer.fit_transform(cars_updated_data)
transformed_x
````

````python
# we have got our data as num 
#now lets fix the model
np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(transformed_x , y , test_size=0.2)
model = RandomForestRegressor()
model.fit(x_train , y_train)
````

````python
model.score(x_test , y_test)
````

###  Two Ways to Make Predictions in Scikit-Learn

1. **`predict()`**  
   - Returns the **final predicted class labels** (for classification) or **values** (for regression).  
   - Example:
     ```python
     y_preds = model.predict(X_test)
     print(y_preds)
     ```

2. **`predict_proba()`**  
   - Returns the **probability estimates** for each class (only for classification models that support it).  
   - Useful when you need confidence levels or want to apply a threshold.
   - Example:
     ```python
     y_probs = model.predict_proba(X_test)
     print(y_probs)
     ```

# **ROC / AUC**

**ROC (Receiver Operating Characteristic)** is a curve that plots the **True Positive Rate (TPR)** vs. the **False Positive Rate (FPR)** at different classification thresholds.

It is used to evaluate the performance of classification models, especially in binary classification tasks.

---

### In our context:
- **Positive class = 1** (e.g., disease present, fraud, etc.)
- **Negative class = 0** (e.g., no disease, not fraud, etc.)

---

### **AUC (Area Under the Curve)**  
AUC represents the degree or measure of separability between the positive and negative classes.  
A higher AUC indicates a better performing model.

---

### 🔑 Key Metrics:

- **True Positive Rate (TPR)** / *Recall*:  
  Proportion of actual positives (1s) correctly identified.  
  `TPR = TP / (TP + FN)`

- **False Positive Rate (FPR)**:  
  Proportion of actual negatives (0s) incorrectly predicted as positives.  
  `FPR = FP / (FP + TN)`

- **True Negative Rate (TNR)** / *Specificity*:  
  Proportion of actual negatives (0s) correctly identified.  
  `TNR = TN / (TN + FP)`

- **False Negative Rate (FNR)**:  
  Proportion of actual positives (1s) incorrectly predicted as negatives.  
  `FNR = FN / (FN + TP)`

---

**Goal:** A good model has a high TPR and low FPR.  
The ROC curve should hug the top-left corner.  
AUC closer to **1** means better performance.

````python
y_probs = clf.predict_proba(x_test)
y_probs[:10] , len(y_probs)
````

````python
y_probs_positive = y_probs[:,1]
y_probs_positive[:10]
````

````python
# calculate fpr , tpr and thresholds

#so basicaly we took test values of 0 and 1 and then compared it with pobability of 1

from sklearn.metrics import roc_curve
fpr , tpr , thresholds = roc_curve(y_test , y_probs_positive)
````

````python
fpr
````

````python
tpr
````

````python
# create a function for plotting ROC curves
import matplotlib.pyplot as plt

def plot_roc(x , y):
    plt.plot(x , y , label = "Roc-Curve")
    # plt.xlabel("False Positive Rate")
    # plt.ylabel("True Positive Rate")
    plt.title("ROC Curve showing fpr and tpr")
    # plt.plot([0,1] , [0,1] , color="red",linestyle="--" , label="guessing")
    plt.legend()
    plt.show()
    

plot_roc(fpr , tpr)
````

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
heart_data = pd.read_csv("heart-disease (1).csv")
np.random.seed(42)
x = heart_data.drop("target" , axis =1)
y = heart_data["target"]
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2)
clf = RandomForestClassifier()
clf.fit(x_train , y_train)
````

````python
y_pred = clf.predict(x_test)
a=confusion_matrix(y_test, y_pred)
a
````
